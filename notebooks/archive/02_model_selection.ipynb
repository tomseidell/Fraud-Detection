# **Model Selection**
This Notebook tests multiple different Machine Learning Algos against our given Problem.


This should give us a first overview and indication towards which models will perform best.


Additionally we will perform Benchmarks, such as RandomGuessing and ZeroGuessing in order to have a real world comparison


The different models we will test are:
- Random Forest
- Decision Tree
- XGBoost 
- LightGBM

## Evaluation Metrics
To evaluate and compare our model we will use typical classification metrics:
- Recall
- Precision
- F1
- ROC-AUC (chooses indpendent threshold for classification. Simply meassures if true has higher % outcome than false)

# **Load merged Dataset**
this Dataset consists of 3 subests:
- X_train, y_train (70%)
- X_val, y_val (15%)
- X_test, y_test (15%)

```python
import sys
sys.path.append('..')  
from src.data.loader import Loader
from src.data.splitter import split_data
from src.models.random_guessing import RandomGuessing
from src.models.zero_guessing import ZeroGuessing
from src.evaluation.evaluation_metrics import EvaluationMetrics

# model dependencies:
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier


loader = Loader().load()
X_train, X_test, y_train, y_test= split_data(loader.df)
```

## Benchmarks
This section will test our Benchmarks 
- RandomGuessing : guesses random number between 0 and 1
- ZeroGuessing : always guesses 0 (no fraud)


based on our evaluation matrics against our given dataset

```python
model = RandomGuessing()
y_pred_test = model.predict_many(X=X_test)
random_guessing_metrics = EvaluationMetrics(y=y_test, y_pred=y_pred_test)
random_guessing_metrics.print_eval_report(headline="RandomGuessing")
```

```python
model = ZeroGuessing()
y_pred_test = model.predict_many(X=X_test)
zero_guessing_metrics = EvaluationMetrics(y=y_test, y_pred=y_pred_test)
zero_guessing_metrics.print_eval_report(headline="ZeroGuessing")
```

## Models
This Section will test the listed models from above

### Decision Tree

```python
model = DecisionTreeClassifier(random_state=42).fit(X_train, y_train)
y_pred_test = model.predict_proba(X_test)[:, 1] # return 1d array of probabilities y=1
metrics = EvaluationMetrics(y=y_test, y_pred=y_pred_test)
metrics.print_eval_report(headline="DecisionTreeClassifier")
```

### Random Forest

```python
model = RandomForestClassifier(random_state=42).fit(X_train, y_train)
y_pred_test = model.predict_proba(X_test)[:, 1]
metrics = EvaluationMetrics(y=y_test, y_pred=y_pred_test)
metrics.print_eval_report(headline="RandomForestClassifier")
```

```python
print(X_train.shape)
```

### XGBoost

```python
model = XGBClassifier(random_state=42).fit(X_train, y_train, eval_set = [(X_val,y_val)])
y_pred_test = model.predict_proba(X_test)[:, 1]
metrics = EvaluationMetrics(y=y_test, y_pred=y_pred_test)
metrics.print_eval_report(headline="XGBClassifier")
```

### LightGBM

```python
model = LGBMClassifier(random_state=42).fit(X_train, y_train, eval_set = [(X_val,y_val)])
y_pred_test = model.predict_proba(X_test)[:, 1]
metrics = EvaluationMetrics(y=y_test, y_pred=y_pred_test)
metrics.print_eval_report(headline="LGBMClassifier")
```

# Summary

All four models were tested on raw, unencoded data (no feature engineering), which sets a performance floor for what the final pipeline needs to beat.

**Decision Tree** failed outright — sklearn's implementation requires numeric input and cannot handle raw string columns. This alone ruled it out.

**Random Forest** and **LightGBM** achieved a ROC-AUC (~0.924–0.925).

**XGBoost** was selected for further development because it handles mixed data types and NaN values natively, which makes sense in a setup with very much missing values
